```python
from pandas import DataFrame as df
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('score68.csv', encoding='utf-8')
```

```python
# Split the dataset into training and validation sets
np.random.seed(0)
np.random.shuffle(df.values)
n = len(df)
train_df, val_df = df[:int(0.8 * n)], df[int(0.8 * n):]
train_df, val_df = train_df.iloc[:,1:], val_df.iloc[:,1:]
print("Total samples:", len(df))
print("Training samples:", len(train_df))
print("Validation samples:", len(val_df))

train_df.to_csv('train.csv', index=False, encoding='utf-8')
val_df.to_csv('validation.csv', index=False, encoding='utf-8')
```

```python
# Split the dataset into training and validation sets use sklearn

from sklearn.model_selection import train_test_split
# Split: 80% train, 20% validation
train_df, val_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

print("Total samples:", len(df))
print("Training samples:", len(train_df))
print("Validation samples:", len(val_df))

train_df.to_csv('train.csv', index=False, encoding='utf-8')
val_df.to_csv('validation.csv', index=False, encoding='utf-8')
```

```python
# Split the dataset into training and validation sets use sklearn and convert output to digit

from sklearn.model_selection import train_test_split

# Split: 80% train, 20% validation
train_df, val_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

# Grade to digit mapping
grade_to_digit = {
    'A': 0,
    'B+': 1,
    'B': 2,
    'C+': 3,
    'C': 4,
    'D+': 5,
    'D': 6,
    'F': 7
}

# Convert the last column to digits
train_df['output'] = train_df.iloc[:, -1].map(grade_to_digit)
val_df['output'] = val_df.iloc[:, -1].map(grade_to_digit)

train_df.to_csv('train.csv', index=False, encoding='utf-8')
val_df.to_csv('validation.csv', index=False, encoding='utf-8')
```

```python
for val in train_df.iloc[:, -1]:
    if val not in grade_to_digit.keys():
        print("Invalid value in training set:", val)
```
