```python
import matplotlib.pyplot as plt
import numpy as np
import pandas
```

```python
iris = pandas.read_csv('IRIS/iris.data')
# iris = pandas.read_csv('IRIS/iris.data', header=None)
columns = ['sepal length','sepal width', 'petal length', 'petal width', 'class']
iris = pandas.read_csv('IRIS/iris.data', header=None, names=columns)
x = np.array(iris)
np.random.shuffle(x)
xn = x[:,:4]
y = x[:,-1]

# x = iris[columns[:-1]]
# y = iris['class']
print(y)
```

```python
# plot sepal length between 3 classes
plt.figure(figsize=(10, 6))

iris_classes = iris['class'].unique()
# 'petal width' "petal length" 'sepal width' 'sepal length'
fid = 3
colors=[[1,0,0],[0,1,0],[0,0,1]]
markers = ["o","s","*"]
for i, species in enumerate(iris_classes):
    species_indx = x[y == species]
    species_indx = ((y == species)*1).nonzero()
    plt.plot(species_indx, x[species_indx,fid], 
             marker=markers[i], linestyle='', color=colors[i], label=species)
```

```python
iris_classes = iris['class'].unique()
iris_classes = {cls:val for val, cls in enumerate(iris_classes)}
print(iris_classes)
# print(np.asarray(y))
```

```python
classes = np.array(list(map(lambda x: iris_classes[x], y)))
# classes = []
# for classname in y:
#     classes.append(iris_classes[classname])
print(classes)
```

```python
# plot data scattering
xn = np.array(iris)
cls1, cls2, cls3 = 0, 1, 2
feat1, feat2 = 0, 3
plt.scatter(xn[classes == cls1, feat1], xn[classes == cls1, feat2],
            color='red', label=f'Class {cls1+1}')
plt.scatter(xn[classes == cls2, feat1], xn[classes == cls2, feat2],
            color='green', label=f'Class {cls2+1}')
plt.scatter(xn[classes == cls3, feat1], xn[classes == cls3, feat2],
            color='blue', label=f'Class {cls3+1}')

plt.xlabel(f'Feature {feat1}')
plt.ylabel(f'Feature {feat2}')
plt.title('Decision Boundary for Single Layer Perceptron')
plt.legend()
plt.show()
```
